特徵順序：Close、MA20、RSI_14、MACD_Hist、Market_Vol_20d、Daily_Sentiment。

預測值是百分點。`0.677330` 代表 +0.677330%。

```
from pathlib import Path
import json, hashlib
root = Path.cwd()
while not (root / 'models' / 'model_555_v2' / 'weights' / 'weights.npz').exists():
    if root.parent == root:
        raise SystemExit('請在交付套件目錄內執行')
    root = root.parent
weight = root / 'models' / 'model_555_v2' / 'weights' / 'weights.npz'
card = json.loads((weight.parent / 'model_card.json').read_text(encoding='utf-8'))
digest = hashlib.sha256(weight.read_bytes()).hexdigest()
expected = '53a1467fdba66e7f22a45e5f2fbc1cb78fc99825b15d516b3e7153d7bfc28144'
print(card['model_version'], digest == expected, digest)
print(card['features'])
```
